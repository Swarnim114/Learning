```python
# Our single neuron function (manual calculation)
import math # Needed for sigmoid

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

def single_neuron(features_list, weights, bias):
    outputs = [] # Where we store the neuron's output for each house

    # Go through each house's features one by one
    for one_house_features in features_list:
        
        # --- Manual Calculation of 'z' ---
        weighted_sum = 0
        # Loop through each feature value and its corresponding weight
        for i in range(len(one_house_features)):
            feature_value = one_house_features[i]
            weight_value = weights[i]
            weighted_sum += (feature_value * weight_value) # Multiply and add to sum
            
        current_z = weighted_sum + bias # Add the bias at the end
        # --- End Manual Calculation ---
        
        # Apply the sigmoid function to 'z' to get the final output (0 to 1)
        current_y_hat = sigmoid(current_z)
        
        outputs.append(current_y_hat) # Add this house's output to our list

    return outputs # Give back all the outputs for each house

# --- Sample Inputs (same as before) ---
all_houses_features = [
    [1500, 3],  # House A
    [800, 1],   # House B
    [500, 1],   # House C
    [1000, 2]   # House D
]

neuron_weights = [0.005, 0.8]
neuron_bias = -2.0

# --- Call the function and print results ---
predictions = single_neuron(all_houses_features, neuron_weights, neuron_bias)
print("Likelihood of houses being expensive (manual calculation):")
print(predictions)
```

```python
# Load dataset
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True)

# Split data
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y)

# Train OvR model
from sklearn.linear_model import LogisticRegression
from sklearn.multiclass import OneVsRestClassifier
ovr = OneVsRestClassifier(LogisticRegression())
ovr.fit(X_train, y_train)

# Evaluate
accuracy = ovr.score(X_test, y_test)
print(f"Accuracy: {accuracy:.2f}")
```
